```python
def validate_user_input(input_data):
    try:
        user_input = UserInput(**input_data)

        print("✅ Valid user input created:")
        print(user_input.model_dump_json(indent=2))

        return user_input

    except ValidationError as e:
        print("❌ Validation error occurred:")

        for error in e.errors():
            print(f" - {error['loc'][0]}: {error['msg']}")

        return None
```

```python
from pydantic import BaseModel, ValidationError, EmailStr, Field
from typing import Optional
from datetime import date
```

```python
class UserInput(BaseModel):
    name: str
    email: EmailStr
    query: str

    order_id: Optional[int] = Field(
        None,
        description="5-digit order number (cannot start with 0)",
        ge=10000,
        le=99999
    )

    purchase_date: Optional[date] = None
```

```python
#Additional Fields Example

input_data = {
    "name": "John Bamigbade",
    "email": "john.o.bamigbade@gmail.com",
    "query": """
I bought a laptop carrying case and it turned out to be
the wrong size. I need to return it.
""",
    "order_id": 12345,
    "purchase_date": date(2025, 12, 31)
}

user_input = validate_user_input(input_data)
```

```python
from pydantic import BaseModel, ValidationError, EmailStr, Field
from typing import Optional, List
from datetime import date
import json
```

```python
# Tool Calling Example

class FAQLookupArgs(BaseModel):
    query: str = Field(
        ...,
        description="User's query"
    )

    tags: List[str] = Field(
        ...,
        description="Relevant keywords/tags from the customer query"
    )
```

```python
# FAQ Lookup Function

def lookup_faq_answer(args: FAQLookupArgs) -> str:
    """
    Look up an FAQ answer by matching tags and words
    in query to FAQ entry keywords.
    """

    faq_database = {
        "password": "Here's a password reset link!",
        "account access": "Try resetting your account credentials."
    }

    for tag in args.tags:
        if tag in faq_database:
            return faq_database[tag]

    return "Please contact support."
```

```python
# Tool Definition

tool_definition = [
    {
        "type": "function",
        "function": {
            "name": "lookup_faq_answer",
            "description": "Look up an FAQ answer by matching tags and words in query to FAQ entry keywords.",
            "parameters": FAQLookupArgs.model_json_schema()
        }
    }
]
```
